```python
# 1. 라이브러리 불러오기
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


# 2. 데이터 로드
diabetes = load_diabetes()
X = diabetes.data       # 10개의 특성(age, sex, bmi, bp, s1~s6)
y = diabetes.target     # 당뇨병 진행도(정수형 연속값)


# 3. 학습 / 테스트 데이터 분리(80:20 비율)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)


# 4. 선형 회귀 모델 생성 및 학습
model = LinearRegression()
model.fit(X_train, y_train)
print(dir(model))


# 5. 예측
y_pred = model.predict(X_test)



# 6. 평가지표 출력
mse = mean_squared_error(y_test, y_pred)    # mse 값이 작아야 모델 학습능률이 좋음.
print(f'mse:{mse}')

r2 = r2_score(y_test, y_pred)
print(f'r2: {r2}')


# 7. 회귀 계수 확인
print(len(diabetes.feature_names))
print(f'weight:{model.coef_}')
print(f'bais:{model.intercept_}')


```
