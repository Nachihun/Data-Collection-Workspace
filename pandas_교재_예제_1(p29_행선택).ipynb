```python
import pandas as pd

exam_data = {'수학' : [90, 80, 70], '영어' : [98, 89, 95],
            '음악' : [85, 95, 100], '체육' : [100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)

df2 = df.copy()
df2 = df2.drop('우현')
print(df2)
```

```python
df3 = df.copy()
df3 = df3.drop(['우현', '인아'], axis=0)
print(df3)
```

```python
df4 = df.copy()
df4 = df.drop(['우현', '인아'], axis='index')
print(df4)
```

```python
df5 = df.copy()
df5 = df.drop(index = ['우현'])
print(df5)
```

```python
import pandas as pd

exam_data = {'수학' : [90, 80, 70], '영어' : [98, 89, 95],
            '음악' : [85, 95, 100], '체육' : [100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)

df2 = df.copy()
df2 = df.drop('수학', axis=1)
print(df2)
```

```python
import pandas as pd

exam_data = {'수학' : [90, 80, 70], '영어' : [98, 89, 95],
            '음악' : [85, 95, 100], '체육' : [100, 90, 90]}

df = pd.DataFrame(exam_data, index=['서준', '우현', '인아'])
print(df)
```

```python
label1 = df.loc['서준']
position1 = df.iloc[0]
print(label1)
print('\n')
print(position1)
```

```python
label2 = df.loc[['서준', '우현']]
position2 = df.iloc[[0, 1]]
print(label2)
print('\n')
print(position2)
```

```python
label3 = df.loc['서준':'우현']
position3 = df.iloc[0:1]
print(label3)
print('\n')
print(position3)
```
