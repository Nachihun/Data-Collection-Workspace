```python

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# 1. 샘플 장바구니 데이터 (트랜잭션 데이터)

dataset = [
['우유', '빵', '기저귀'],
['우유', '기저귀', '맥주', '계란'],
['빵', '기저귀', '맥주'],
['우유', '빵', '기저귀', '맥주'],
['빵', '우유', '계란'],
['빵', '모자', '가위', '사과'],
['맥주', '소주', '콜라', '사이다']
]

# 2. 원-핫 인코딩 매트릭스 변환 (True/False 형태)

te = TransactionEncoder()
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

# 3. Apriori 알고리즘으로 빈발 항목집합(Frequent Itemsets) 추출

# min_support=0.4 : 전체 거래의 40% 이상 등장하는 항목만 선별

frequent_itemsets = apriori(df, min_support=0.35, use_colnames=True)

# 4. 연관 규칙(Association Rules) 생성

# metric="lift", min_threshold=1.0 : 향상도가 1.0 이상인 규칙만 추출

rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

# 5. 보기 쉽게 주요 컬럼 정렬 및 출력

result = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
result = result.sort_values(by='lift', ascending=False)

print(result.to_string(index=False))
```
